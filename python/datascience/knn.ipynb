```python
import numpy as np
import pandas as pd

data={
'Name':['Sara' , 'Ali' , 'Mina','Reza','Neda','Amir','Hoda'] ,
'Age':[21,23,np.nan,22,24,21,23],
'GPA':[18.5,np.nan,19.2,14.5,20,85,17.5],
'City':['Tehran','Shiraz','Tehran','Isfahan',np.nan,'Tehran','Shiraz'],
'Education':['Bachelor','Bachelor','Master','Bachelor','Master','Bachelor',np.nan]

}
df=pd.DataFrame(data)
print(df)
```

```python
print(f"Shape: {df.shape}")
print(f" \nColumns: {df.columns}")
print(f"\nData Types: \n{df.dtypes}")
print(f"\nDataFrame Info:")
df.info()

```

```python
print(df.isnull().sum())
```

```python
age_mean=df['Age'].mean()
df['Age']=df['Age'].fillna(age_mean)
print(df['Age'])
```

```python
gpa_median=df['GPA'].median()
df['GPA']=df['GPA'].fillna(gpa_median)
print(df['GPA'])
```

```python
city_mode=df['City'].mode()[0]
df['City']=df['City'].fillna(city_mode)

Education_mode=df['Education'].mode()[0]
df['Education']=df['Education'].fillna(Education_mode)

print(df[['City','Education']])
```

```python
Q1=df['GPA'].quantile(0.25)
Q3=df['GPA'].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5 * IQR
upper_bound=Q3+1.5 * IQR
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers=df[
    (df['GPA']< lower_bound) |
    (df['GPA']> upper_bound)
]

print("\nOutliers:")
print(outliers)
```

```python
df=df[
    (df['GPA'] >= lower_bound) &
    (df['GPA'] <= upper_bound)
]

df=df.reset_index(drop=True)
print(df)
```

```python
print(f"cleaned dataframe: \n{df}")
print("\nMissing values:")
print(df.isnull().sum())
```
